```python
# Create a copy; preserve the original raw dataframe
charging_clean = charging_df.copy()

# Broad geographic screening bounds for India
valid_coordinates = (
    charging_clean["latitude"].between(6, 38) &
    charging_clean["longitude"].between(68, 98)
)

# A record is eligible only if its source quality is OK
charging_clean["geo_validated"] = (
    charging_clean["location_quality"].eq("ok") &
    valid_coordinates
)

# Record the reason for validation outcome
charging_clean["geo_validation_reason"] = "Passed"

charging_clean.loc[
    charging_clean["location_quality"].eq("suspect"),
    "geo_validation_reason"
] = "Source flagged as suspect"

charging_clean.loc[
    ~valid_coordinates,
    "geo_validation_reason"
] = "Coordinates outside screening bounds"

# Display validation summary
print("Total records:", len(charging_clean))
print("\nGeographic validation:")
print(charging_clean["geo_validated"].value_counts())

print("\nValidation reasons:")
print(charging_clean["geo_validation_reason"].value_counts())
```

```python
import ast

structured_columns = [
    "chargers",
    "standards",
    "currents",
    "vehicle_classes"
]

for col in structured_columns:
    charging_clean[col] = charging_clean[col].apply(
        lambda x: ast.literal_eval(x) if isinstance(x, str) else x
    )

print("Updated data types:")

for col in structured_columns:
    print(col, ":", charging_clean[col].apply(type).value_counts().to_dict())
```

```python
charging_clean["charger_list_count"] = (
    charging_clean["chargers"].apply(
        lambda x: len(x) if isinstance(x, list) else None
    )
)

charging_clean["charger_count_match"] = (
    charging_clean["charger_list_count"] ==
    charging_clean["total_chargers"]
)

print(charging_clean["charger_count_match"].value_counts(dropna=False))

print("\nMismatched records:")
display(
    charging_clean.loc[
        ~charging_clean["charger_count_match"],
        ["id", "total_chargers", "charger_list_count"]
    ].head(10)
)
```

```python
# Step 1: Calculate actual charger count from the nested charger details

charging_clean["calculated_chargers"] = (
    charging_clean["chargers"].apply(
        lambda x: sum(
            item.get("count", 1)
            for item in x
            if isinstance(item, dict)
        )
    )
)

# Step 2: Compare calculated count with total_chargers

charging_clean["charger_count_match"] = (
    charging_clean["calculated_chargers"] ==
    charging_clean["total_chargers"]
)

# Step 3: Display validation results

print("Charger Count Validation:")
print(charging_clean["charger_count_match"].value_counts())

# Step 4: Display remaining mismatches

mismatches = charging_clean[
    ~charging_clean["charger_count_match"]
]

print("\nRemaining mismatches:", len(mismatches))

display(
    mismatches[
        ["id", "total_chargers", "calculated_chargers", "chargers"]
    ].head(10)
)
```

```python
# Remove temporary validation columns
charging_clean.drop(
    columns=[
        "charger_list_count",
        "charger_count_match",
        "calculated_chargers"
    ],
    errors="ignore",
    inplace=True
)

# Check duplicate station IDs
print(
    "Duplicate station IDs:",
    charging_clean["id"].duplicated().sum()
)

# Check duplicate rows using station ID
print(
    "Duplicate station records:",
    charging_clean.duplicated(subset=["id"]).sum()
)

# Dataset shape
print("\nFinal dataset shape:", charging_clean.shape)

# Geographic validation
print("\nGeographic validation:")
print(charging_clean["geo_validated"].value_counts())

# Missing values
print("\nMissing values:")
print(charging_clean.isnull().sum().sum())
```

```python
# Create state-name mapping

state_mapping = {
    "Andaman and Nicobar Islands": "Andaman & Nicobar Island",
    "Dadra and Nagar Haveli and Daman and Diu": "UT of DNH and DD"
}

# Create copies so original cleaned datasets remain unchanged

statewise_merge = statewise_clean.copy()
charging_merge = charging_clean.copy()

# Standardize charging dataset state names

charging_merge["state"] = charging_merge["state"].replace(state_mapping)

# Validate state names again

registration_states = set(statewise_merge["State"].str.strip())
charging_states = set(charging_merge["state"].str.strip())

print("Registration States:", len(registration_states))
print("Charging States:", len(charging_states))

print("\nStates only in Registration Dataset:")
print(sorted(registration_states - charging_states))

print("\nStates only in Charging Dataset:")
print(sorted(charging_states - registration_states))

print("\nDo both datasets now have matching state names?",
      registration_states == charging_states)
```

```python
# Aggregate charging infrastructure by state

charging_state_summary = (
    charging_merge.groupby("state")
    .agg(
        Total_Stations=("id", "nunique"),
        Total_Chargers=("total_chargers", "sum"),
        Total_Connectors=("total_connectors", "sum")
    )
    .reset_index()
)

# Rename state column for merging

charging_state_summary.rename(
    columns={"state": "State"},
    inplace=True
)

print("Charging State Summary Shape:", charging_state_summary.shape)

display(
    charging_state_summary.sort_values(
        by="Total_Stations",
        ascending=False
    ).head(10)
)
```

```python
# Merge registration and charging infrastructure datasets

ev_infrastructure = pd.merge(
    statewise_merge,
    charging_state_summary,
    on="State",
    how="inner",
    validate="one_to_one"
)

# Check merged dataset

print("Merged Dataset Shape:", ev_infrastructure.shape)

print("\nMerged Dataset:")
display(ev_infrastructure.head(10))

print("\nMissing Values:")
print(ev_infrastructure.isnull().sum())
```

```python
# Calculate EV registrations per charging station

ev_infrastructure["EVs_Per_Station"] = (
    ev_infrastructure["EV_Registrations"] /
    ev_infrastructure["Total_Stations"]
)

# Sort states by highest EV-to-station ratio

ev_infrastructure = ev_infrastructure.sort_values(
    by="EVs_Per_Station",
    ascending=False
).reset_index(drop=True)

display(
    ev_infrastructure[
        [
            "State",
            "EV_Registrations",
            "Total_Stations",
            "EVs_Per_Station"
        ]
    ].head(10)
)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

ev_infrastructure["Demand_Score"] = (
    scaler.fit_transform(
        ev_infrastructure[["EV_Registrations"]]
    ) * 100
).round(2)

ev_infrastructure["Infrastructure_Pressure_Score"] = (
    scaler.fit_transform(
        ev_infrastructure[["EVs_Per_Charger"]]
    ) * 100
).round(2)

display(
    ev_infrastructure[
        [
            "State",
            "EV_Registrations",
            "EVs_Per_Charger",
            "Demand_Score",
            "Infrastructure_Pressure_Score"
        ]
    ].sort_values(
        by="Infrastructure_Pressure_Score",
        ascending=False
    ).head(10)
)
```

```python
# Calculate preliminary infrastructure screening score

ev_infrastructure["Infrastructure_Gap_Score"] = (
    0.50 * ev_infrastructure["Demand_Score"] +
    0.50 * ev_infrastructure["Infrastructure_Pressure_Score"]
).round(2)

# Rank states by screening score

ev_infrastructure["Gap_Rank"] = (
    ev_infrastructure["Infrastructure_Gap_Score"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

# Display top 15 states

gap_analysis = ev_infrastructure.sort_values(
    by="Infrastructure_Gap_Score",
    ascending=False
)

display(
    gap_analysis[
        [
            "State",
            "EV_Registrations",
            "Total_Stations",
            "Total_Chargers",
            "EVs_Per_Charger",
            "Demand_Score",
            "Infrastructure_Pressure_Score",
            "Infrastructure_Gap_Score",
            "Gap_Rank"
        ]
    ].head(15)
)
```

### Preliminary Infrastructure Gap Analysis

1. Uttar Pradesh ranks first with an infrastructure screening score of 82.18.
2. Bihar ranks second with 66.03, reflecting high EV-to-charger pressure.
3. Assam ranks third with 60.90 and has the highest EV-to-charger ratio.
4. Chandigarh ranks fourth, but its relatively small EV market requires additional consideration.
5. Maharashtra ranks fifth despite having the second-highest EV registrations because it has a comparatively large charging network.

**Business Implication:**
Uttar Pradesh, Bihar and Assam are priority candidates for deeper infrastructure assessment. Chandigarh should be evaluated separately because its small geographic size and EV base may influence its score.

**Methodology:**
The preliminary score uses equal weights: 50% normalized EV demand and 50% normalized EV-to-charger pressure.

**Limitation:**
This score is a screening tool, not a verified measure of charging shortages or investment profitability. Registration and charging datasets have different reference dates. Geographic distribution, charger power, compatibility, utilization, accessibility, land and grid costs are not included.

```python
# Numerical outlier analysis

numeric_columns = [
    "total_chargers",
    "total_connectors",
    "max_power_kw"
]

outlier_summary = []

for column in numeric_columns:

    Q1 = charging_clean[column].quantile(0.25)
    Q3 = charging_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = charging_clean[
        (charging_clean[column] < lower_bound) |
        (charging_clean[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / len(charging_clean) * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary)
```

### Numerical Outlier Analysis – Insights

1. The IQR method identified 3,498 records as statistical outliers in total_chargers.
2. It identified 6,760 records in total_connectors.
3. It identified 4,455 records in max_power_kw.
4. The first two columns have an IQR of zero because both Q1 and Q3 equal 1.
5. Therefore, the IQR method over-flags records above 1 for these two columns.
6. Higher charging power values may represent genuine high-capacity charging infrastructure.

**Data Quality Decision:**
No records were removed based solely on IQR results. Unusual values require contextual validation before being classified as errors.

```python
# Calculate EV-to-charger and EV-to-station ratios
ev_infrastructure["EVs_Per_Charger"] = (
    ev_infrastructure["EV_Registrations"] / ev_infrastructure["Total_Chargers"]
)

ev_infrastructure["EVs_Per_Station"] = (
    ev_infrastructure["EV_Registrations"] / ev_infrastructure["Total_Stations"]
)

# Calculate normalized demand and infrastructure pressure scores
ev_infrastructure["Demand_Score"] = (
    (ev_infrastructure["EV_Registrations"] - ev_infrastructure["EV_Registrations"].min()) /
    (ev_infrastructure["EV_Registrations"].max() - ev_infrastructure["EV_Registrations"].min()) * 100
)

ev_infrastructure["Infrastructure_Pressure_Score"] = (
    (ev_infrastructure["EVs_Per_Charger"] - ev_infrastructure["EVs_Per_Charger"].min()) /
    (ev_infrastructure["EVs_Per_Charger"].max() - ev_infrastructure["EVs_Per_Charger"].min()) * 100
)

# Preliminary screening score
ev_infrastructure["Infrastructure_Gap_Score"] = (
    0.5 * ev_infrastructure["Demand_Score"] +
    0.5 * ev_infrastructure["Infrastructure_Pressure_Score"]
)

print("Calculated columns added successfully.")
display(ev_infrastructure.head())
```
