```python

import requests
import pandas as pd
import time
import json

base_url = "https://ev-charging-stations-api.vercel.app/v1/stations"

all_stations = []
page = 1
limit = 100

session = requests.Session()

while True:

    params = {
        "page": page,
        "limit": limit
    }

    response = session.get(
        base_url,
        params=params,
        timeout=60
    )

    if response.status_code == 429:
        print("Rate limit reached. Waiting 15 minutes...")
        time.sleep(910)
        continue

    response.raise_for_status()

    result = response.json()

    records = result["data"]

    if not records:
        break

    all_stations.extend(records)

    print(f"Page {page}: {len(records)} records collected")

    if page >= result["meta"]["total_pages"]:
        break

    page += 1

    # Respect the API's rate limit
    time.sleep(10)

print("Total records:", len(all_stations))

# Preserve original API records
with open(
    "EV_Charging_Stations_Raw.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(all_stations, file, ensure_ascii=False, indent=2)

# Create station-level CSV
df_stations = pd.json_normalize(all_stations)

df_stations.to_csv(
    "EV_Charging_Stations_Raw.csv",
    index=False
)

print("CSV saved successfully!")
display(df_stations.head())
```
